# CHATBOT WITH LANGCHAIN FAISSDB OPENAI LLM

## Step 1 : Import all the required Softwares

```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
import faiss
import os
from langchain_community.vectorstores import FAISS
from openai import OpenAI
```

## Step 2 : Load the tcs pdf file and Load all the pages

```python
loader = PyPDFLoader('tcs.pdf')
mydocuments = loader.load()
```

## Step 3 - Convert documents to chunks

```python
text_splitter = RecursiveCharacterTextSplitter(chunk_size=2500,chunk_overlap=500)
mychunks = text_splitter.split_documents(mydocuments)
```

## Step 4 - Take a Vector Model

```python
model = HuggingFaceEmbeddings(model_name ='sentence-transformers/all-mpnet-base-v2')
```

## Step 5 - Create faiss db and pass documents and Vector model

```python
vector_store = FAISS.from_documents(mychunks, model)
```

## Step 6 - Ask a question on tcs pdf get relevant chunks from faiss db

```python
results = vector_store.similarity_search(
    "How is TCS performing in terms of revenue?",
    k=3
)
```

## Step 7 - Keep the all 3 chunks results into one variable

```python
our_context= ''
for res in results:
    our_context += res.page_content+'\n\n'
```

## Step 8 - Make a prompt

```python
our_prompt = f"""
I am going to ask a question based on my context.
Please answer the question based on my context only.
If you don't find the answer in my context,
tell me you did not find the answer, and don't give me
an answer from your own knowledge.

context = {our_context}

question = {"How is TCS performing in terms of revenue?"}
"""
```

## Step 9 - Give Prompt to OpenAI LLM and Get Response

```python
import os
os.environ['OPENAI_API_KEY']="OPENAI_API_KEY"
client = OpenAI()
response = client.responses.create(model = "gpt-5.6-luna",input = " who is Dhoni?")

print(response.output_text)
```
